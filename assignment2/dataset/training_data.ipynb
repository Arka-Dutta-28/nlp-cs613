```python
bpcc['label'] = bpcc['tgt_lang'].map(label_to_lang)
bpcc['script'] = bpcc['tgt_lang'].map(label_to_script)
```

```python
ba['script'] = ba['script'].replace(
    {'Perso-Arabic': 'Arabic', 'Perso_Arabic': 'Arabic'}
)
```

```python
bpcc.drop(["src_lang", "tgt_lang", "src"], axis=1, inplace=True)
bpcc = bpcc.rename(columns={'tgt': 'native', 'Roman': 'roman'})
```

```python
from datasets import load_dataset
import random

langs = ["hi", "bn", "ta", "ml", "te", "gu", "mr", "pa", "or", "kn", "as"]
english_sentences = []

for lang in langs:
    try:
        print(f"Loading {lang} split from Samanantar...")
        ds = load_dataset("ai4bharat/samanantar", lang, split="train[:200000]")

        # Filter sentences that are strings and have between 10 and 50 words
        filtered = [
            row["src"] for row in ds
            if isinstance(row["src"], str)
            and 10 <= len(row["src"].split()) <= 50
        ]

        english_sentences.extend(filtered)
        print(f"Loaded {len(filtered):,} valid English sentences for {lang}")

    except Exception as e:
        print(f"Skipping {lang} due to error: {e}")

print(f"Total English sentences collected: {len(english_sentences):,}")

# Sanity check
if not english_sentences:
    raise RuntimeError("No English sentences collected! Check dataset structure or filtering criteria.")

# Randomly assign one English sentence to each row
bpcc["english_random_sentence"] = random.choices(
    english_sentences, k=len(bpcc)
)

print("Added 'english_random_sentence' column from Samanantar dataset.")
```

```python
ba.drop("source", axis=1, inplace=True)
```

```python
combined = pd.concat([ba, bpcc], axis=0, ignore_index=True)
combined.head()
```

```python
combined.to_csv("combined.csv", index=False)
```

```python
sampled_dfs = []

for lang, group in combined.groupby("label"): 
    sampled = group.sample(n=100_345, random_state=42)
    sampled_dfs.append(sampled)

combined_balanced = pd.concat(sampled_dfs, ignore_index=True)
print("Samples per language (after undersampling):")
print(combined_balanced["label"].value_counts())

# Calculate size in bytes → MB / GB
df_size_bytes = combined_balanced.memory_usage(deep=True).sum()
df_size_mb = df_size_bytes / (1024 ** 2)
df_size_gb = df_size_bytes / (1024 ** 3)

print(f"\nTotal samples: {len(combined_balanced):,}")
print(f"Languages included: {combined_balanced['label'].nunique()}")
print(f"DataFrame size: {df_size_mb:.2f} MB ({df_size_gb:.3f} GB)")
```

```python
combined_balanced = combined_balanced[['native','roman','english_random_sentence']].rename(
    columns={'english_random_sentence': 'english'})
combined_balanced.to_csv("triplet_nre.csv", index=False)
print("NRE(type A) triplets are saved")
```

## Creating triplets with Native pair of Same language, Native of different language

```python
# =============== Sampling 100k sentences for each class from same dataframe =============

sampled_dfs = []

for lang, group in combined.groupby("label"): 
    sampled = group.sample(n=100_345, random_state=24)
    sampled_dfs.append(sampled)

combined_balanced = pd.concat(sampled_dfs, ignore_index=True)
print(combined_balanced["label"].value_counts())
```

```python
df = combined_balanced[['native', 'label']]
df.head()
```

```python
import pandas as pd
import numpy as np
from tqdm import tqdm

tqdm.pandas()

# --- Step 1: Group once by label (fast) ---
class_sentences = {
    label: group['native'].values
    for label, group in tqdm(df.groupby('label'), desc="Grouping sentences by label")
}

labels = list(class_sentences.keys())

# --- Step 2: Compute triplet counts per label (stratified like stratify=y) ---
class_counts = df['label'].value_counts(normalize=True).to_dict()
total_triplets = len(df)
triplets_per_class = {
    label: int(round(class_counts[label] * total_triplets))
    for label in labels
}

# --- Step 3: Generate stratified + nested-balanced triplets ---
triplet_chunks = []

for label in tqdm(labels, desc="Generating balanced nested triplets by class"):
    sentences = class_sentences[label].copy()
    np.random.shuffle(sentences)

    total_needed = triplets_per_class[label]
    n_other = len(labels) - 1
    per_negative_label = total_needed // n_other
    remainder = total_needed % n_other

    # ensure even pairing for A-B
    if len(sentences) < 2:
        continue

    # Repeat if not enough to cover all pairs
    needed_pairs = total_needed
    if len(sentences) < needed_pairs * 2:
        sentences = np.tile(sentences, int(np.ceil((needed_pairs * 2) / len(sentences))))
    if len(sentences) % 2 != 0:
        sentences = sentences[:-1]

    # Create large shuffled A/B pool
    A_pool = sentences[::2]
    B_pool = sentences[1::2]

    ptr = 0
    chunks_per_label = []

    for neg_label in labels:
        if neg_label == label:
            continue

        # Determine how many pairs to take for this negative label
        n_samples = per_negative_label + (1 if remainder > 0 else 0)
        remainder -= 1 if remainder > 0 else 0

        # Slice A/B pairs from the shuffled pool
        A = A_pool[ptr:ptr + n_samples]
        B = B_pool[ptr:ptr + n_samples]
        ptr += n_samples

        # Sample negatives (C) from this specific negative label
        neg_sentences = class_sentences[neg_label]
        C = np.random.choice(neg_sentences, size=n_samples, replace=len(neg_sentences) < n_samples)

        # Combine this chunk
        chunk = np.column_stack([A, B, C, np.full(len(A), label, dtype=object)])
        chunks_per_label.append(chunk)

    # Merge chunks for this label
    if chunks_per_label:
        triplet_chunks.append(np.vstack(chunks_per_label))

# --- Step 4: Combine all triplets ---
triplet_array = np.vstack(triplet_chunks)
triplet_df = pd.DataFrame(triplet_array, columns=['A', 'B', 'C', 'label'])

# --- Step 5: Shuffle and trim ---
triplet_df = triplet_df.sample(frac=1, random_state=42).reset_index(drop=True)
triplet_df = triplet_df.iloc[:len(df)]

# --- Step 6: Verify class and negative balance ---
print(f"\nGenerated {len(triplet_df):,} balanced triplets (same as original rows)")
print("Triplet anchor class distribution:")
print(triplet_df['label'].value_counts(normalize=True).round(3))

# Build a mapping from sentence -> label (fast)
sentence_to_label = {}
for label, sentences in class_sentences.items():
    for s in sentences:
        sentence_to_label[s] = label

# Map C sentences to labels quickly
neg_series = triplet_df['C'].map(sentence_to_label)

print("\nNegative sentence class distribution (should be roughly uniform per anchor):")
print(neg_series.value_counts(normalize=True).round(3))

triplet_df.head()
```

```python
# Count duplicate pairs (A, B) per class
duplicate_pair_counts = triplet_df.groupby('label').apply(
    lambda g: pd.Series(zip(g['A'], g['B'])).duplicated().sum(), include_groups=False
)

# Percentage of duplicates per class
duplicate_pair_pct = triplet_df.groupby('label').apply(
    lambda g: pd.Series(zip(g['A'], g['B'])).duplicated().mean() * 100, include_groups=False
)

# Combine results
dup_pair_df = pd.DataFrame({
    'dup_pairs_count': duplicate_pair_counts,
    'dup_pairs_pct': duplicate_pair_pct
})

print(dup_pair_df)
```
